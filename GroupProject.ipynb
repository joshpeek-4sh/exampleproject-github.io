```python
import pandas as pd
import numpy as np

# Import the original dataset
df = pd.read_csv("bean-there-coffee.csv")

# Create a copy to preserve the original dataset
clean = df.copy()

# 1. Remove duplicate rows
clean = clean.drop_duplicates()

# 2. Standardize store names
clean["store"] = (
    clean["store"].str.strip().str.lower()
    .replace({
        "campus store": "campus",
        "down town": "downtown",
        "airport t2": "airport"
    })
    .str.title()
)

# 3. Convert timestamp into datetime format
clean["timestamp"] = pd.to_datetime(clean["timestamp"])

# 4. Handle missing values
clean["size"] = clean["size"].fillna("No size")
clean["discount_code"] = clean["discount_code"].fillna("None")

# 5. Calculate total sales
clean["total_sales"] = clean["unit_price"] * clean["quantity"]

# Display cleaned dataset
display(clean.head())

# Check cleaning results
print("Original rows:", len(df))
print("Cleaned rows:", len(clean))
print("Remaining duplicates:", clean.duplicated().sum())
print("Missing values:")
print(clean.isnull().sum())
```

```python
# Export cleaned dataset as a new CSV file
clean.to_csv("bean-there-coffee-cleaned.csv", index=False)

print("Cleaned CSV successfully created!")
```
